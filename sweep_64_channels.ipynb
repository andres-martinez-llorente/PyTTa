```python
import os

# Set environment variable before importing sounddevice. Value is not important.
os.environ["SD_ENABLE_ASIO"] = "1"

import pytta
```

```python

device = pytta.get_device_from_user()

sweep = pytta.generate.sweep(samplingRate= 48000, fftDegree=18)
sweep.plot_time()

# sin = pytta.generate.sin(samplingRate=48000,fftDegree=18)
# sin.plot_time()
```

```python
measurementParams = {
    'excitation': sweep,
    'samplingRate': 48000,
    'freqMin': 20,
    'freqMax': 20000,
    'device': device,
    'inChannels': list(range(1,65)),
    'outChannels': list(range(1,65)),
    'comment': 'Testing; 1, 2.'
}

ms = pytta.generate.measurement('playrec',
                                **measurementParams)

med1 = ms.run()
med1.plot_time()
```
